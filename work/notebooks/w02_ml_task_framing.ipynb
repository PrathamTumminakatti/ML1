## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

Task Type: Ranking / Scoring

Reason: The business decision is to determine which content items should be refreshed first. Editors act on a prioritized list rather than a simple yes/no prediction. Therefore, ranking/scoring is the appropriate ML task because it produces an ordered priority score for each content item, enabling efficient allocation of editorial effort.


## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target: A priority score that ranks content items by how urgently they should be refreshed.

Label Source: The target should come from observed future performance (such as a measured decline in traffic or search performance in a later time window), not from a defined rule. I would not use is_declining_label, trend_direction, or trend_pct as the target because they are rule-derived and can introduce data leakage. Using an observed outcome ensures the model learns real-world patterns rather than reproducing an existing rule.

```python
import pandas as pd
from google.colab import files

# Upload dataset
uploaded = files.upload()

# Load the uploaded file automatically
filename = next(iter(uploaded))
df = pd.read_csv(filename)

print("Dataset loaded successfully.")
print("Shape:", df.shape)

# Candidate target/proxy columns
candidate_columns = [
    "is_declining_label",
    "trend_direction",
    "trend_pct"
]

# Keep only columns that actually exist
available_columns = [col for col in candidate_columns if col in df.columns]

if available_columns:
    print("\nAvailable target/proxy columns:")
    print(available_columns)

    print("\nSample values:")
    print(df[available_columns].head())

    for col in available_columns:
        print(f"\nValue counts for {col}:")
        print(df[col].value_counts(dropna=False))
else:
    print("\nNo predefined target/proxy columns found.")
    print("Available columns are:")
    print(df.columns.tolist())
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Success Metric: Precision@10

Good Result: A Precision@10 of 0.80 or higher, meaning at least 8 out of the top 10 content items recommended by the model genuinely require refreshing based on observed future performance. This metric is appropriate because the goal is to help editors prioritize a small number of high-impact pages rather than evaluate every page equally.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule cannot reliably identify which content should be refreshed because content performance depends on the interaction of many factors, such as traffic, click-through rate, engagement, search position, freshness, and content type. These relationships are complex and vary across different pages and clients. A machine learning model can learn these patterns from historical data and produce a more accurate priority ranking than a simple if-else rule based on one or two thresholds.

## 2. The question: decision, action, cost of a wrong call

## The Question: Decision, Action, and Cost of a Wrong Call

This project aims to improve the decision of **which content pages should be reviewed or refreshed first**. The recommendations are intended for **SEO specialists and content editors**, who can use them to prioritize their work and focus on pages that are most likely to decline in performance. A wrong recommendation has two possible costs. A **false positive** may cause editors to spend time updating content that did not actually need attention, wasting time and resources. A **false negative** may fail to identify genuinely declining content, allowing it to continue losing search visibility and traffic before any action is taken.

## Careful Words: What I Can and Can't Claim

This project will identify **observed patterns** and **directional relationships** between historical SEO and engagement metrics and declining content performance.
